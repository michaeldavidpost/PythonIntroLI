### Continue

```python
for letter, animals in animalLookup.items():
    if len(animals) > 1:
        continue
    print(f'Only one animal! {animals}')
```

```python
for letter, animals in animalLookup.items():
    if len(animals) > 1:
        continue
    print(f'Only one animal! {animals[0]}')
```

### Break

```python
for letter, animals in animalLookup.items():
    if len(animals) > 1:
        print (f'Found {len(animals)} animals: {animals}')
        break
```

### For / Else

```python
for number in range(2, 100):
    for factor in range(2, int(number**0.5) + 1):
        if number % factor == 0:
            break
    else:
        print(f'{number} is prime!')
```

```python
for number in range(2,100):
    found_factors = False
    for factor in range(2, int(number**0.5) + 1):
        if number % factor == 0:
            found_factors = True
            break
    if not found_factors:
        print(f'{number} is prime!')
```
